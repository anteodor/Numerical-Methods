## Mathematical Background

Numerical integration is a fundamental area of numerical analysis concerned with approximating the value of definite integrals. Such methods are particularly useful when an analytical solution is difficult to obtain or when the function is available only at discrete points. The general problem can be formulated as the evaluation of the definite integral

$$
I = \int_a^b f(x)\,dx.
$$

The exact value of the integral represents the area under the curve $f(x)$ over the interval $[a,b]$. Numerical integration methods approximate this area by replacing the curve with simpler geometric shapes such as rectangles or trapezoids.

### Analytical Solution of the Integral

For the function

$$
f(x)=x^2,
$$

defined on the interval

$$
[1,4],
$$

the exact value of the integral can be obtained analytically.

First, the antiderivative of the function is determined:

$$
\int x^2\,dx = \frac{x^3}{3} + C.
$$

Applying the Newton–Leibniz formula gives

$$
\int_1^4 x^2\,dx
=
\left[\frac{x^3}{3}\right]_1^4
=
\frac{4^3}{3}-\frac{1^3}{3}.
$$

Substituting the limits yields

$$
\int_1^4 x^2\,dx
=
\frac{64}{3}-\frac{1}{3}
=
\frac{63}{3}
=
21.
$$

Therefore, the exact value of the integral is

$$
I_{\text{exact}} = 21.
$$

This value is used as a reference solution for evaluating the accuracy of the numerical methods implemented in this laboratory work.
### Integral of the Gaussian Function

In the second part of the laboratory work, the following integral is considered:

$$
I = \int_{-1}^{3} e^{-x^2}\,dx.
$$

Unlike the polynomial function considered previously, the function

$$
f(x)=e^{-x^2}
$$

does not possess an elementary antiderivative. Therefore, the integral cannot be evaluated analytically using standard elementary functions.

The antiderivative of the Gaussian function is expressed through the error function:

$$
\int e^{-x^2}\,dx
=
\frac{\sqrt{\pi}}{2}\,\mathrm{erf}(x)+C,
$$

where

$$
\mathrm{erf}(x)
=
\frac{2}{\sqrt{\pi}}
\int_0^x e^{-t^2}\,dt.
$$

Consequently, the exact analytical evaluation of

$$
\int_{-1}^{3} e^{-x^2}\,dx
$$

requires the use of special functions and is not straightforward.

For this reason, the integral is approximated numerically using the rectangle and trapezoidal methods implemented in this laboratory work. A highly accurate reference value is obtained using the `scipy.integrate.quad` function and is used to assess the accuracy of the numerical approximations.

The reference value computed by SciPy is approximately

$$
\int_{-1}^{3} e^{-x^2}\,dx
\approx
1.633031.
$$

This value serves as the benchmark for the error analysis performed in Task 3.

### Left Rectangle Method

The left rectangle method approximates the integral by evaluating the function at the left endpoint of each subinterval. For a partition consisting of $n$ subintervals, the approximation is given by

$$
I_L \approx
\sum_{i=0}^{n-1}
f(x_i)\,\Delta x,
$$

where

$$
\Delta x = \frac{b-a}{n}.
$$

Since the height of each rectangle is determined by the function value at the left boundary, this method tends to underestimate increasing functions and overestimate decreasing functions.

### Right Rectangle Method

The right rectangle method uses the function value at the right endpoint of each subinterval:

$$
I_R \approx
\sum_{i=1}^{n}
f(x_i)\,\Delta x.
$$

For increasing functions, this method typically produces an overestimation of the integral.

### Midpoint Rectangle Method

The midpoint rectangle method evaluates the function at the center of each subinterval:

$$
I_M \approx
\sum_{i=1}^{n}
f\left(\frac{x_{i-1}+x_i}{2}\right)\Delta x.
$$

Because the midpoint provides a more representative value of the function over each interval, this method generally achieves higher accuracy than the left and right rectangle methods.

### Trapezoidal Method

The trapezoidal method approximates the area under the curve by replacing each segment of the function with a straight line. The area of each subinterval is therefore represented by a trapezoid:

$$
I_T \approx
\sum_{i=1}^{n}
\frac{f(x_{i-1})+f(x_i)}{2}\,\Delta x.
$$

This method typically provides a more accurate approximation than the basic rectangle methods because it incorporates information from both endpoints of each subinterval.

### Relative Error

To evaluate the accuracy of the numerical approximations, the relative error is calculated as

$$
\varepsilon =
\left|
\frac{I_{\text{approx}}-I_{\text{exact}}}
     {I_{\text{exact}}}
\right|
\times 100\%.
$$

A smaller relative error indicates a more accurate numerical approximation. The calculated errors allow a direct comparison of the effectiveness of the implemented integration methods.

## 3. Program Implementation

The laboratory work was implemented in Python using the NumPy, Matplotlib, and SciPy libraries. The implementation was organized into several modular functions responsible for computing exact integral values, performing numerical approximations, visualizing the results, and evaluating approximation accuracy.

A dedicated function was created to compute a highly accurate reference value of a definite integral using the `scipy.integrate.quad` routine. This value serves as a benchmark for assessing the performance of the implemented numerical methods.

The numerical integration procedure was implemented in a universal function that supports multiple approximation techniques through a method-selection parameter. The implementation includes the Left Rectangle Method, Right Rectangle Method, Midpoint Rectangle Method, and Trapezoidal Method. The integration interval is automatically divided into equally spaced subintervals, and the corresponding approximation is calculated by summing the contributions of individual segments.

To evaluate the quality of the numerical approximations, a separate function was implemented for calculating the relative error with respect to the reference solution. This enables a quantitative comparison of the accuracy achieved by different numerical integration methods.

For visualization purposes, an additional plotting function was developed using Matplotlib. The function displays the integrated curve together with the geometric approximation corresponding to the selected numerical method. Such visual representations provide a clear illustration of how each technique estimates the area under the curve.

The program was applied to two test functions. The first function, $f(x)=x^2$, was used to compare numerical approximations with a known analytical solution. The second function, $f(x)=e^{-x^2}$, was selected to investigate the behavior of numerical integration methods for a function without an elementary antiderivative. The developed implementation provides a flexible framework for evaluating and comparing numerical integration methods while maintaining a clear and modular program structure.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import integrate
```

```python
# Function from Task 1
def f1(x):
    """
    Function f(x) = x^2.
    """
    return x**2


# Function from Task 3
def f2(x):
    """
    Function f(x) = exp(-x^2).
    """
    return np.exp(-x**2)
```

```python
def exact_integral(func, a, b):
    """
    Calculates the exact/reference value of the integral
    using scipy.integrate.quad.
    """
    if a > b:
        raise ValueError("The upper limit b must be greater than a.")

    result, error = integrate.quad(func, a, b)
    return result
```

```python
def numerical_integral(func, a, b, nodes, method="middle"):
    """
    Calculates a numerical approximation of the definite integral.
    """

    if nodes < 2:
        raise ValueError("The number of nodes must be at least 2.")

    if a > b:
        raise ValueError("The upper limit b must be greater than a.")

    x = np.linspace(a, b, nodes)
    integral_sum = 0

    for i in range(1, len(x)):
        left = x[i - 1]
        right = x[i]
        h = right - left

        if method == "left":
            integral_sum += h * func(left)

        elif method == "right":
            integral_sum += h * func(right)

        elif method == "middle":
            midpoint = (left + right) / 2
            integral_sum += h * func(midpoint)

        elif method == "trapezoidal":
            integral_sum += h * (func(left) + func(right)) / 2

        else:
            raise ValueError("Unknown method. Use: left, right, middle, trapezoidal.")

    return integral_sum
```

```python
def relative_error(approximation, exact):
    """
    Calculates the relative error in percent.
    """
    return abs((approximation - exact) / exact) * 100
```

```python
def plot_numerical_integration(func, a, b, nodes, method="middle"):
    """
    Plots the function and the corresponding numerical approximation.
    """

    x_nodes = np.linspace(a, b, nodes)
    x_dense = np.linspace(a, b, 500)

    plt.figure(figsize=(7, 5))
    plt.plot(x_dense, func(x_dense), color="red", label="Function")

    for i in range(1, len(x_nodes)):
        left = x_nodes[i - 1]
        right = x_nodes[i]

        if method == "left":
            height = func(left)
            plt.plot([left, left], [0, height], color="blue")
            plt.plot([right, right], [0, height], color="blue")
            plt.plot([left, right], [height, height], color="blue")

        elif method == "right":
            height = func(right)
            plt.plot([left, left], [0, height], color="blue")
            plt.plot([right, right], [0, height], color="blue")
            plt.plot([left, right], [height, height], color="blue")

        elif method == "middle":
            midpoint = (left + right) / 2
            height = func(midpoint)
            plt.plot([left, left], [0, height], color="blue")
            plt.plot([right, right], [0, height], color="blue")
            plt.plot([left, right], [height, height], color="blue")

        elif method == "trapezoidal":
            plt.plot([left, left], [0, func(left)], color="blue")
            plt.plot([right, right], [0, func(right)], color="blue")
            plt.plot([left, right], [func(left), func(right)], color="blue")

    plt.title(f"Numerical approximation: {method}")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.grid(True)
    plt.legend()
    plt.show()
```

```python
# TASK 1–2
# Integral of f(x) = x^2 on [1, 4]
# Using 3 intervals means 4 nodes: x = 1, 2, 3, 4

a1 = 1
b1 = 4
nodes1 = 4

exact1 = exact_integral(f1, a1, b1)

methods = {
    "left": "Left rectangles",
    "right": "Right rectangles",
    "middle": "Middle rectangles",
    "trapezoidal": "Trapezoidal method"
}

print("TASK 1–2: Integral of f(x) = x^2 on [1, 4]")
print("Exact value:", exact1)
print("-" * 70)

results_task1 = []

for method, name in methods.items():
    approximation = numerical_integral(f1, a1, b1, nodes1, method)
    error = relative_error(approximation, exact1)

    results_task1.append([name, approximation, error])

    print(f"{name}:")
    print(f"Approximation = {approximation}")
    print(f"Relative error = {error:.6f}%")
    print("-" * 70)

    plot_numerical_integration(f1, a1, b1, nodes1, method)
```

```python
# Displaying results in a table

import pandas as pd

task1_table = pd.DataFrame(
    results_task1,
    columns=["Method", "Approximation", "Relative error (%)"]
)

task1_table
```

```python
# TASK 3
# Integral of exp(-x^2) on [-1, 3]
# Comparison of accuracy for different numbers of nodes

a2 = -1
b2 = 3

exact2 = exact_integral(f2, a2, b2)

nodes_list = [5, 10, 20, 50, 100]

print("TASK 3: Integral of exp(-x^2) on [-1, 3]")
print("Reference value:", exact2)
print("-" * 70)

results_task3 = []

for nodes in nodes_list:
    for method, name in methods.items():
        approximation = numerical_integral(f2, a2, b2, nodes, method)
        error = relative_error(approximation, exact2)

        results_task3.append([nodes, name, approximation, error])

task3_table = pd.DataFrame(
    results_task3,
    columns=["Nodes", "Method", "Approximation", "Relative error (%)"]
)

task3_table
```

```python
# Plotting exp(-x^2) with midpoint rectangles for visualization

plot_numerical_integration(f2, a2, b2, nodes=10, method="middle")
```

```python
# Error dependence on the number of nodes

plt.figure(figsize=(8, 5))

for method, name in methods.items():
    method_data = task3_table[task3_table["Method"] == name]
    plt.plot(
        method_data["Nodes"],
        method_data["Relative error (%)"],
        marker="o",
        label=name
    )

plt.title("Dependence of relative error on the number of nodes")
plt.xlabel("Number of nodes")
plt.ylabel("Relative error (%)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Final comparison table sorted by error

task3_table.sort_values(by="Relative error (%)")
```

## 4. Results and Discussion

The implemented numerical integration methods were successfully applied to both test functions considered in this laboratory work. The obtained results demonstrate the influence of the selected numerical method and the number of integration nodes on the accuracy of the approximation.

For the polynomial function $f(x)=x^2$, the numerical approximations were compared with the exact analytical solution. As expected, the Left Rectangle Method underestimated the value of the integral, while the Right Rectangle Method produced an overestimation. This behavior is a direct consequence of the monotonic increase of the function over the integration interval. Since the function values continuously increase, selecting the left endpoint of each subinterval leads to smaller rectangle heights, whereas selecting the right endpoint results in larger heights.

The Midpoint Rectangle Method and the Trapezoidal Method produced considerably more accurate approximations. Both methods provide a better representation of the function within each subinterval and therefore reduce the approximation error. The obtained results confirm the theoretical expectation that these methods achieve higher accuracy than the basic left and right rectangle approaches.

For the function $f(x)=e^{-x^2}$, the numerical integration was performed using different numbers of nodes in order to investigate the convergence properties of the implemented methods. The results clearly indicate that increasing the number of nodes improves the approximation accuracy for all methods. As the discretization becomes finer, the numerical estimate follows the shape of the function more closely, reducing the integration error.

Among the considered methods, the Midpoint Rectangle Method consistently provided the most accurate approximations throughout the experiments. The Trapezoidal Method also demonstrated excellent performance and rapidly converged toward the reference solution as the number of nodes increased. In contrast, the Left and Right Rectangle Methods exhibited larger errors, particularly when a small number of nodes was used. Nevertheless, their accuracy improved steadily with increasing discretization density.

The error convergence plot further confirms these observations. For all methods, the relative error decreases as the number of nodes increases. However, the rate of convergence differs among the methods. The Midpoint Rectangle Method and the Trapezoidal Method converge significantly faster, indicating their superior efficiency for practical numerical integration tasks.

Overall, the conducted experiments validate the correctness of the implemented algorithms and demonstrate the fundamental principles of numerical integration. The results show that both the choice of integration method and the number of nodes play a crucial role in determining the accuracy of the approximation. Among the tested approaches, the Midpoint Rectangle Method provided the best balance between simplicity and accuracy, while the Trapezoidal Method also proved to be a reliable and efficient numerical integration technique.
